```python

import pandas as pd

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
edges = pd.read_csv("txs_edgelist.csv")
sample = pd.read_csv("sample_submission.csv")

for name, df in [
    ("TRAIN", train),
    ("TEST", test),
    ("EDGES", edges),
    ("SUBMISSION", sample)
]:
    print(f"\n=== {name} ===")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist()[:10])
    print("Duplicates:", df.duplicated().sum())
    print("Missing values:", df.isna().sum().sum())
    print(df.head(3).to_string())

print("\n=== LABEL DISTRIBUTION ===")
print(train["label"].value_counts(dropna=False))

print("\n=== TIME STEPS ===")
for name, df in [("Train", train), ("Test", test)]:
    print(
        name,
        "Min:", df["time_step"].min(),
        "Max:", df["time_step"].max(),
        "Unique:", df["time_step"].nunique()
    )

print("\n=== LABELS BY TIME STEP ===")
print(
    pd.crosstab(
        train["time_step"],
        train["label"],
        dropna=False
    )
)

print("\n=== TRANSACTION ID CHECK ===")
print("Train unique IDs:", train["txId"].nunique())
print("Test unique IDs:", test["txId"].nunique())
print(
    "Train/test overlapping IDs:",
    len(set(train["txId"]) & set(test["txId"]))
)

print("\n=== EDGE COVERAGE ===")
nodes = set(train["txId"]) | set(test["txId"])
print("Total graph edges:", len(edges))
print("Graph nodes represented:",
      len(set(edges.iloc[:, 0]) | set(edges.iloc[:, 1])))
print("Edges with both endpoints in available nodes:",
      (edges.iloc[:, 0].isin(nodes) &
       edges.iloc[:, 1].isin(nodes)).sum())

print("\n=== SUBMISSION FORMAT ===")
print(sample.shape)
print(sample.columns.tolist())
print(sample.head())
```
